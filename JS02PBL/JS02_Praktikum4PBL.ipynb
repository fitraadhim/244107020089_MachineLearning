# JS02 Praktikum 4 PBL - Visualisasi Dataset Sleep Wise

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv(Path('../Dataset/Sleep_health_and_lifestyle_dataset.csv'))
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None')
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['Sleep Disorder'].value_counts().plot(kind='bar', ax=axes[0], color=['#4c72b0', '#dd8452', '#55a868'])
axes[0].set_title('Distribusi Sleep Disorder')
axes[0].set_xlabel('Kelas')
axes[0].set_ylabel('Jumlah')
for label, group in df.groupby('Sleep Disorder'):
    axes[1].scatter(group['Sleep Duration'], group['Quality of Sleep'], label=label, alpha=0.7)
axes[1].set_title('Durasi dan Kualitas Tidur')
axes[1].set_xlabel('Sleep Duration')
axes[1].set_ylabel('Quality of Sleep')
axes[1].legend()
plt.tight_layout()
plt.show()
```
