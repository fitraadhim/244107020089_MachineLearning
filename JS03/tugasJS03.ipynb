```python
import pandas as pd

df = pd.read_csv('wbc.csv')

print("5 Data Pertama:")
display(df.head())
print("\nInformasi Data:")
df.info()
```

```python
# Biasanya kolom ID atau kolom kosong/tidak relevan dibuang jika ada.
# Berdasarkan struktur umum dataset WBC, kolom 'id' tidak digunakan untuk prediksi,
# dan kolom 'diagnosis' adalah target/label.

if 'id' in df.columns:
    df_model = df.drop(columns=['id'])
    print("Kolom 'id' berhasil dihapus karena tidak digunakan.")
else:
    df_model = df.copy()
    print("Tidak ada kolom 'id', menggunakan seluruh kolom.")

# Memisahkan Fitur (X) dan Target (y)
X = df_model.drop(columns=['diagnosis'])
y = df_model['diagnosis']

print("\nFitur (X) shape:", X.shape)
print("Target (y) shape:", y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

# Mengubah label 'M' (Malignant) dan 'B' (Benign) menjadi nilai numerik (0 dan 1)
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Menampilkan hasil encoding unik
print("Mapping kelas asli ke numerik:", dict(zip(le.classes_, le.transform(le.classes_))))
```

```python
from sklearn.preprocessing import StandardScaler

# Melakukan standarisasi pada semua kolom fitur numerik (X)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Mengembalikan ke bentuk DataFrame agar mudah dilihat
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print("Hasil Standarisasi (5 baris pertama):")
display(X_scaled_df.head())
```

```python
from sklearn.model_selection import train_test_split

# Melakukan stratified split 80% data latih dan 20% data uji berdasarkan target
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df,
    y_encoded,
    test_size=0.2,
    stratify=y_encoded,
    random_state=42
)

print(f"Total data latih: {X_train.shape[0]}")
print(f"Total data uji: {X_test.shape[0]}")
print("\nDistribusi kelas pada data latih (proporsi stratified):")
print(pd.Series(y_train).value_counts(normalize=True))
```
