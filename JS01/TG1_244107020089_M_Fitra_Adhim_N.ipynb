1. Contoh Tindakan Melanggar Etika dan Hukum dalam Penggunaan AI
- Contoh Pelanggaran Etika: Generasi Deepfake Non-Konsensual dan Misinformasi
  
  Bentuk Pelanggaran: Penggunaan model AI generatif (seperti GANs atau model diffusion) untuk merekayasa wajah atau suara seseorang tanpa izin, baik untuk menyebarkan berita bohong (hoaks) politik maupun membuat konten tidak etis.

  Alasan: Melanggar prinsip etika kecerdasan buatan seperti keadilan (fairness), privasi, transparansi, serta berpotensi merusak reputasi dan rasa aman individu.

- Contoh Pelanggaran Hukum: Pelanggaran Hak Cipta (Copyright Infringement) Data Pelatihan
  
  Bentuk Pelanggaran: Pengikisan (scraping) karya seni, artikel, atau kode program berhak cipta secara masif tanpa izin pemilik hak cipta untuk melatih model AI komersial (Large Language Models atau Text-to-Image).

  Alasan & Regulasi: Kasus hukum ini melanggar undang-undang hak cipta di berbagai negara (misalnya UU Hak Cipta No. 28 Tahun 2014 di Indonesia atau gugatan fair use di tingkat internasional).

2. Dampak Energi dan Lingkungan Pemanfaatan AI & Solusinya
- Dampak Energi dan Lingkungan

  Konsumsi Listrik & Jejak Karbon: Proses pelatihan (training) dan inference model AI berskala besar (seperti LLM atau model pemrosesan data medis/sinyal) membutuhkan daya komputasi GPU/TPU yang sangat tinggi di data center. Hal ini memicu emisi karbon yang signifikan jika sumber energinya masih berbasis bahan bakar fosil.

  Konsumsi Air (Water Footprint): Data center memerlukan jutaan liter air bersih untuk sistem pendingin (cooling system) server selama proses komputasi berlangsung.

- Cara Mengatasi (Solusi & Pendapat Pribadi)

  Implementasi Green AI & Efisiensi Algoritma: Mengalihkan fokus riset dari sekadar memperbesar ukuran model (Red AI) ke optimasi efisiensi model (Green AI), seperti menerapkan teknik prunisasi (pruning), kuantisasi (quantization), dan distilasi pengetahuan (knowledge distillation) agar model dapat berjalan dengan konsumsi daya lebih rendah.

  Transisi ke Energi Terbarukan: Perusahaan penyedia infrastruktur cloud dan data center wajib menggunakan sumber energi terbarukan (seperti panel surya atau angin) serta sistem pendingin tertutup (closed-loop cooling) yang hemat air.

  Pendapat Pribadi: Sebagai pengembang/praktisi AI, kita harus bersikap bijak dengan tidak selalu melatih model dari awal (train from scratch) jika sudah tersedia pre-trained model yang relevan (transfer learning). Selain itu, pemilihan arsitektur model harus disesuaikan dengan kebutuhan nyata—tidak semua masalah memerlukan model deep learning yang sangat besar jika algoritma yang lebih ringan sudah mencukupi.

Referensi:

- Strubell, E., Ganesh, A., & McCallum, A. (2019). Energy and Policy Considerations for Deep Learning in NLP. Proceedings of the 57th Annual Meeting of the Association for Computational Linguistics (ACL).
- Wu, C. J., et al. (2022). Sustainable AI: Environmental Implications, Challenges and Opportunities. Proceedings of MLSys.
