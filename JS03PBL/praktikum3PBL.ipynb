# JS03 Praktikum 3 PBL - Random Split

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv(Path('../Dataset/Sleep_health_and_lifestyle_dataset.csv'))
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None')
train, test = train_test_split(df, test_size=0.2, random_state=42, stratify=df['Sleep Disorder'])
print('Data training:', train.shape)
print('Data testing:', test.shape)
print('Distribusi training:\n', train['Sleep Disorder'].value_counts())
print('Distribusi testing:\n', test['Sleep Disorder'].value_counts())
```
