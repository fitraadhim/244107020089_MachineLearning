# JS05 Praktikum 2 PBL - Evaluasi Density Clustering

Notebook ini merupakan implementasi PBL menggunakan dataset *Sleep Health and Lifestyle*.

## 1. Tujuan

Mengevaluasi parameter eps dan min_samples pada DBSCAN menggunakan silhouette dan rasio noise.

## 2. Dataset dan Alur Praktikum

Dataset disimpan pada `Dataset/Sleep_health_and_lifestyle_dataset.csv`. Alur: **load -> cleaning -> feature engineering -> pemodelan -> evaluasi -> visualisasi**.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

plt.style.use("seaborn-v0_8-whitegrid")

csv_path = Path("../Dataset/Sleep_health_and_lifestyle_dataset.csv")
df = pd.read_csv(csv_path)
print("Ukuran dataset:", df.shape)
df.head()
```

## 3. Eksplorasi dan Pembersihan Data

```python
print("Informasi dataset:")
df.info()

print("\nNilai kosong:")
print(df.isna().sum())
print("\nDuplikasi:", df.duplicated().sum())

df = df.copy()
df["Sleep Disorder"] = df["Sleep Disorder"].fillna("None")

bp = df["Blood Pressure"].str.split("/", expand=True).astype(float)
df["Systolic BP"] = bp[0]
df["Diastolic BP"] = bp[1]
print("\nDistribusi Sleep Disorder:")
print(df["Sleep Disorder"].value_counts())
```

### Cek kualitas data dan ringkasan statistik

```python
display(df.describe(include="number").T)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df["Sleep Disorder"].value_counts().plot(kind="bar", ax=axes[0], color="teal")
axes[0].set_title("Distribusi Sleep Disorder")
axes[0].set_xlabel("Kelas")
axes[0].set_ylabel("Jumlah")

df["Sleep Duration"].hist(bins=20, ax=axes[1], color="cornflowerblue", edgecolor="black")
axes[1].set_title("Distribusi Durasi Tidur")
axes[1].set_xlabel("Jam tidur")
axes[1].set_ylabel("Frekuensi")
plt.tight_layout()
plt.show()
```

### Persiapan Fitur untuk Clustering

Membentuk matriks fitur `X` (tanpa kolom identitas, tekanan darah mentah, dan target) serta `preprocessor` untuk encoding kategorikal dan standardisasi numerik sebelum clustering.

```python
drop_columns = ["Person ID", "Blood Pressure", "Sleep Disorder"]
X = df.drop(columns=drop_columns)

numeric_features = X.select_dtypes(include=["number"]).columns.tolist()
categorical_features = [col for col in X.columns if col not in numeric_features]

preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), numeric_features),
    ("categorical", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_features),
])

print("Fitur numerik:", numeric_features)
print("Fitur kategorik:", categorical_features)
print("Ukuran matriks fitur:", X.shape)
```

## 4. Pemodelan dan Eksperimen DBSCAN

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

X_ready = preprocessor.fit_transform(X)
rows = []
for eps in [0.8, 1.0, 1.2, 1.5]:
    for min_samples in [4, 5, 8]:
        labels_try = DBSCAN(eps=eps, min_samples=min_samples).fit_predict(X_ready)
        n_clusters = len(set(labels_try)) - (1 if -1 in labels_try else 0)
        noise = float((labels_try == -1).mean())
        mask = labels_try != -1
        score = silhouette_score(X_ready[mask], labels_try[mask]) if n_clusters > 1 and mask.sum() > n_clusters else float("nan")
        rows.append([eps, min_samples, n_clusters, noise, score])

results = pd.DataFrame(rows, columns=["eps", "min_samples", "n_clusters", "noise_ratio", "silhouette"])
display(results.sort_values(["silhouette", "noise_ratio"], ascending=[False, True]).head(10))

best_row = results.dropna(subset=["silhouette"]).sort_values(
    ["silhouette", "noise_ratio"], ascending=[False, True]
).iloc[0]
labels = DBSCAN(
    eps=float(best_row["eps"]), min_samples=int(best_row["min_samples"])
).fit_predict(X_ready)
print(
    f"DBSCAN terbaik: eps={best_row['eps']}, min_samples={int(best_row['min_samples'])}, "
    f"cluster={int(best_row['n_clusters'])}, noise={best_row['noise_ratio']:.2%}"
)
```

## 5. Visualisasi Cluster dan Noise

```python
from sklearn.decomposition import PCA

X_ready = preprocessor.fit_transform(X)
X_2d = PCA(n_components=2, random_state=42).fit_transform(X_ready)
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels, cmap="tab10",
                      s=35, edgecolor="black")
plt.title("DBSCAN Cluster dan Noise")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar(scatter, label="Label (-1 = noise)")
plt.tight_layout()
plt.show()
```

## 6. Kesimpulan

Nilai eps dan min_samples yang terlalu kecil atau besar dapat menyebabkan banyak noise atau terlalu sedikit cluster.
