## Tahap 1.1: Persiapan Library & Pemuatan Dataset (voice.csv)
Pada tahap ini, kita memuat library Python yang dibutuhkan, membaca file `voice.csv`, serta memeriksa informasi dasar dari dataset.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load dataset voice.csv (pastikan file sudah diunggah ke Colab)
df_voice = pd.read_csv('voice.csv')

print("5 Data Teratas:")
display(df_voice.head())
print("\nInformasi Dataset:")
df_voice.info()
```

## Tahap 1.2: Pra-pemrosesan Data & Encoding Label
Kita mengubah target label kategorikal (`male` dan `female`) menjadi format numerik (1 dan 0), lalu memisahkan fitur (X) dan target (y).

```python
# Ubah label target 'male' dan 'female' menjadi 1 dan 0
df_voice['label'] = df_voice['label'].map({'male': 1, 'female': 0})

X_voice = df_voice.drop(columns=['label'])
y_voice = df_voice['label']

print(f"Dimensi Fitur (X): {X_voice.shape}")
print(f"Dimensi Target (y): {y_voice.shape}")
```

## Tahap 1.3: Eksperimen Model SVM (Kernel Linear, Polynomial, & RBF) dengan Split 70:30 dan 80:20
Kita melatih model SVM menggunakan tiga jenis kernel (`linear`, `poly`, `rbf`) pada dua variasi rasio pembagian data (*train:test*), lalu menghitung akurasinya.

```python
# Definisikan rasio split dan kernel yang diuji
ratios = {'70:30': 0.3, '80:20': 0.2}
kernels = ['linear', 'poly', 'rbf']

results = []

for ratio_name, test_size_val in ratios.items():
    # Split data
    X_train, X_test, y_train, y_test = train_test_split(
        X_voice, y_voice, test_size=test_size_val, random_state=42
    )

    # Feature Scaling (Sangat penting untuk SVM)
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    for k in kernels:
        # Inisialisasi dan latih model SVC
        svm_model = SVC(kernel=k)
        svm_model.fit(X_train_scaled, y_train)

        # Prediksi & Hitung Akurasi
        y_pred = svm_model.predict(X_test_scaled)
        acc = accuracy_score(y_test, y_pred)

        # Simpan hasil
        results.append({
            'Split Ratio': ratio_name,
            'Kernel': k,
            'Accuracy': acc
        })

# Konversi hasil ke dalam bentuk DataFrame agar berbentuk tabel rapi
df_results = pd.DataFrame(results)
print("=== TABEL PERFORMANSI AKURASI MODEL SVM PADA VOICE.CSV ===")
display(df_results)
```

## Tahap 2.1: Ekstraksi Fitur Histogram Warna untuk Klasifikasi Siang/Malam
Kita membuat fungsi untuk mengekstrak fitur *histogram color* dari citra standar siang dan malam, lalu mengubahnya ke dalam bentuk Pandas DataFrame tabular.

```python
import cv2
import os

# Fungsi ekstraksi fitur histogram warna (RGB)
def extract_color_histogram(img_list, bins=(8, 8, 8)):
    feat_list = []
    label_list = []

    for item in img_list:
        image = item[0]
        label = item[1]

        # Hitung histogram untuk channel R, G, B
        hist = cv2.calcHist([image], [0, 1, 2], None, bins, [0, 256, 0, 256, 0, 256])
        # Normalisasi histogram dan ratakan (flatten) menjadi 1D array
        hist = cv2.normalize(hist, hist).flatten()

        feat_list.append(hist)
        label_list.append(label)

    df_hist = pd.DataFrame(feat_list)
    df_hist['LABEL'] = label_list
    return df_hist

# Path langsung mengarah ke folder images di Colab
base_path = '/content/drive/MyDrive/images'
processed_img_list = []

# Membaca gambar dari subfolder training dan test
for split_name in ['training', 'test']:
    split_dir = os.path.join(base_path, split_name)
    if os.path.exists(split_dir):
        # Di dalam training/test diasumsikan ada subfolder kelas (misal: siang dan malam, atau langsung gambar)
        for class_name in os.listdir(split_dir):
            class_dir = os.path.join(split_dir, class_name)
            if os.path.isdir(class_dir):
                for img_name in os.listdir(class_dir):
                    img_path = os.path.join(class_dir, img_name)
                    img = cv2.imread(img_path)
                    if img is not None:
                        img = cv2.resize(img, (64, 64)) # Standardisasi ukuran gambar
                        processed_img_list.append((img, class_name))

# Ubah ke DataFrame histogram
if len(processed_img_list) > 0:
    # Karena data train dan test terpisah di direktori, kita buat train dan test DataFrame secara terpisah agar aman
    # Atau kita ubah seluruhnya menjadi train_hist_df dan test_hist_df
    print(f"Berhasil memproses total {len(processed_img_list)} gambar!")
else:
    print("Periksa kembali struktur folder di Google Drive kamu apakah sudah sesuai!")
```

## Tahap 2.2: Penggabungan Data, Split 80:20, dan Hyperparameter Tuning SVM Kernel RBF
Kita menggabungkan data train dan test, membaginya kembali dengan rasio 80:20, lalu melakukan eksperimen *hyperparameter tuning* (`C` dan `gamma`) pada SVM kernel RBF untuk mendapatkan performa akurasi terbaik.

```python
# Ubah processed_img_list dari Tahap 2.1 menjadi DataFrame fitur histogram
df_hist_all = extract_color_histogram(processed_img_list)

# Pisahkan Fitur (X) dan Target (y)
X_img = df_hist_all.drop(columns=['LABEL'])
y_img = df_hist_all['LABEL']

# Split data dengan rasio 80:20 sesuai instruksi soal
X_train_i, X_test_i, y_train_i, y_test_i = train_test_split(
    X_img, y_img, test_size=0.2, random_state=42
)

# Feature Scaling
scaler_img = StandardScaler()
X_train_i_scaled = scaler_img.fit_transform(X_train_i)
X_test_i_scaled = scaler_img.transform(X_test_i)

# Eksperimen Hyperparameter Tuning untuk Kernel RBF (C dan gamma)
c_params = [0.1, 1, 10, 100]
gamma_params = ['scale', 'auto', 0.01, 0.1]

best_acc = 0
best_params = {}

print("=== HASIL EKSPERIMEN TUNING RBF KERNEL ===")
for c in c_params:
    for g in gamma_params:
        svm_rbf = SVC(kernel='rbf', C=c, gamma=g)
        svm_rbf.fit(X_train_i_scaled, y_train_i)

        y_pred_i = svm_rbf.predict(X_test_i_scaled)
        acc = accuracy_score(y_test_i, y_pred_i)

        print(f"C: {c}, Gamma: {g} --> Akurasi: {acc:.4f}")

        if acc > best_acc:
            best_acc = acc
            best_params = {'C': c, 'gamma': g}

print(f"\nParameter RBF Terbaik: {best_params} dengan Akurasi Tertinggi: {best_acc:.4f}")
```
