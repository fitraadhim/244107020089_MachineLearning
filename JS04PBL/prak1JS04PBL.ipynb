Notebook ini merupakan implementasi PBL menggunakan dataset *Sleep Health and Lifestyle*.

## 1. Tujuan

Memahami cara kerja dan implementasi K-Means untuk segmentasi pola kesehatan tidur.

## 2. Dataset dan Alur Praktikum

Dataset disimpan pada `Dataset/Sleep_health_and_lifestyle_dataset.csv`. Alur: **load -> cleaning -> feature engineering -> pemodelan -> evaluasi -> visualisasi**.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

plt.style.use("seaborn-v0_8-whitegrid")

csv_path = Path("../Dataset/Sleep_health_and_lifestyle_dataset.csv")
df = pd.read_csv(csv_path)
print("Ukuran dataset:", df.shape)
df.head()
```

## 3. Eksplorasi dan Pembersihan Data

```python
print("Informasi dataset:")
df.info()

print("\nNilai kosong:")
print(df.isna().sum())
print("\nDuplikasi:", df.duplicated().sum())

df = df.copy()
df["Sleep Disorder"] = df["Sleep Disorder"].fillna("None")

bp = df["Blood Pressure"].str.split("/", expand=True).astype(float)
df["Systolic BP"] = bp[0]
df["Diastolic BP"] = bp[1]
print("\nDistribusi Sleep Disorder:")
print(df["Sleep Disorder"].value_counts())
```

### Cek kualitas data dan ringkasan statistik

```python
display(df.describe(include="number").T)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df["Sleep Disorder"].value_counts().plot(kind="bar", ax=axes[0], color="teal")
axes[0].set_title("Distribusi Sleep Disorder")
axes[0].set_xlabel("Kelas")
axes[0].set_ylabel("Jumlah")

df["Sleep Duration"].hist(bins=20, ax=axes[1], color="cornflowerblue", edgecolor="black")
axes[1].set_title("Distribusi Durasi Tidur")
axes[1].set_xlabel("Jam tidur")
axes[1].set_ylabel("Frekuensi")
plt.tight_layout()
plt.show()
```

### Persiapan Fitur untuk Clustering

Membentuk matriks fitur `X` (tanpa kolom identitas, tekanan darah mentah, dan target) serta `preprocessor` untuk encoding kategorikal dan standardisasi numerik sebelum clustering.

```python
drop_columns = ["Person ID", "Blood Pressure", "Sleep Disorder"]
X = df.drop(columns=drop_columns)

numeric_features = X.select_dtypes(include=["number"]).columns.tolist()
categorical_features = [col for col in X.columns if col not in numeric_features]

preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), numeric_features),
    ("categorical", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_features),
])

print("Fitur numerik:", numeric_features)
print("Fitur kategorik:", categorical_features)
print("Ukuran matriks fitur:", X.shape)
```

## 4. Pemodelan Clustering

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.pipeline import Pipeline

scores = {}
cluster_labels = {}
for k in range(2, 7):
    pipe = Pipeline([("preprocessor", preprocessor), ("kmeans", KMeans(n_clusters=k, n_init=20, random_state=42))])
    labels = pipe.fit_predict(X)
    X_ready = pipe.named_steps["preprocessor"].transform(X)
    scores[k] = silhouette_score(X_ready, labels)
    cluster_labels[k] = labels

print("Silhouette per jumlah cluster:")
for k, score in scores.items():
    print(f"k={k}: {score:.4f}")

best_k = max(scores, key=scores.get)
labels = cluster_labels[best_k]
print("Jumlah cluster terbaik:", best_k)
```

## 5. Visualisasi Hasil

```python
from sklearn.decomposition import PCA

X_ready = preprocessor.fit_transform(X)
X_2d = PCA(n_components=2, random_state=42).fit_transform(X_ready)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels, cmap="viridis", s=35, edgecolor="black")
plt.title("Visualisasi Cluster dalam Ruang 2D (PCA)")
plt.xlabel("Komponen Utama 1")
plt.ylabel("Komponen Utama 2")
plt.colorbar(scatter, label="Cluster")
plt.tight_layout()
plt.show()

cluster_profile = df.copy()
cluster_profile["Cluster"] = labels
display(cluster_profile.groupby("Cluster")[["Age", "Sleep Duration", "Quality of Sleep",
                                           "Stress Level", "Daily Steps"]].mean().round(2))
```
